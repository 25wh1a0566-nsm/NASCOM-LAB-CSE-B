```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
data = {
    "Student": ["S1", "S2", "S3", "S4", "S5", "S6", "S7", "S8", "S9", "S10"],
    "Study_Hours": [2, 3, 4, 5, 6, 7, 8, 4, 6, 9],
    "Attendance": [65, 70, 75, 80, 85, 90, 92, 78, 88, 95],
    "Assignment_Score": [55, 60, 65, 72, 78, 85, 88, 70, 80, 92],
    "Previous_Score": [58, 62, 68, 70, 76, 82, 86, 72, 79, 90],
    "Final_Score": [60, 64, 70, 75, 82, 88, 91, 74, 84, 94],
}
df = pd.DataFrame(data)

print("=== PART A: BASIC INFORMATION ===")
print("Number of students:", len(df))
print("\nColumn Names:", df.columns.tolist())
print("\nData Types:\n", df.dtypes)
print("\nFirst 5 Records:\n", df.head())
print("\nLast 5 Records:\n", df.tail())

print("\nDataset Info:")
df.info()
```

```python

print("\n=== PART B: PANDAS DATA ANALYSIS ===")

print("\nComplete Dataset:\n", df)
averages = df[
    [
        "Study_Hours",
        "Attendance",
        "Assignment_Score",
        "Previous_Score",
        "Final_Score",
    ]
].mean()
print("\nAverages across all performance metrics:\n", averages)

highest_student = df.loc[df["Final_Score"].idxmax()]
print(
    f"\nHighest Final Score: {highest_student['Student']} ({highest_student['Final_Score']})"
)

lowest_student = df.loc[df["Final_Score"].idxmin()]
print(
    f"Lowest Final Score:  {lowest_student['Student']} ({lowest_student['Final_Score']})"
)
print("\nStudents with Attendance > 80%:\n", df[df["Attendance"] > 80])

print("\nStudents with Final Score > 80:\n", df[df["Final_Score"] > 80])

print("\nStatistical Summary (describe()):\n", df.describe())
```

```python

print("\n=== PART C: NUMPY ANALYSIS ===")

final_scores = df["Final_Score"].to_numpy()

print("Final Scores Array:", final_scores)
print("Mean:              ", np.mean(final_scores))
print("Median:            ", np.median(final_scores))
print("Maximum:           ", np.max(final_scores))
print("Minimum:           ", np.min(final_scores))
print("Standard Deviation:", np.std(final_scores))
```

```python
import matplotlib.pyplot as plt
import pandas as pd

data = {
    "Student": ["S1", "S2", "S3", "S4", "S5", "S6", "S7", "S8", "S9", "S10"],
    "Study_Hours": [2, 3, 4, 5, 6, 7, 8, 4, 6, 9],
    "Attendance": [65, 70, 75, 80, 85, 90, 92, 78, 88, 95],
    "Assignment_Score": [55, 60, 65, 72, 78, 85, 88, 70, 80, 92],
    "Previous_Score": [58, 62, 68, 70, 76, 82, 86, 72, 79, 90],
    "Final_Score": [60, 64, 70, 75, 82, 88, 91, 74, 84, 94],
}

df = pd.DataFrame(data)

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

axes[0, 0].scatter(
    df["Study_Hours"], df["Final_Score"], color="#1f77b4", s=80, edgecolors="k"
)
axes[0, 0].set_title("Study Hours vs Final Score", fontsize=12, fontweight="bold")
axes[0, 0].set_xlabel("Study Hours")
axes[0, 0].set_ylabel("Final Score")

axes[0, 1].scatter(
    df["Attendance"], df["Final_Score"], color="#2ca02c", s=80, edgecolors="k"
)
axes[0, 1].set_title("Attendance vs Final Score", fontsize=12, fontweight="bold")
axes[0, 1].set_xlabel("Attendance (%)")
axes[0, 1].set_ylabel("Final Score")

axes[1, 0].hist(
    df["Final_Score"], bins=5, color="#9467bd", edgecolor="black", alpha=0.85
)
axes[1, 0].set_title(
    "Distribution of Final Scores", fontsize=12, fontweight="bold"
)
axes[1, 0].set_xlabel("Final Score")
axes[1, 0].set_ylabel("Number of Students")

bars = axes[1, 1].bar(
    df["Student"], df["Final_Score"], color="#8c564b", edgecolor="black"
)
bars[df["Final_Score"].idxmax()].set_color("#d62728")
bars[df["Final_Score"].idxmin()].set_color("#ff7f0e")
axes[1, 1].set_title(
    "Student-wise Final Scores", fontsize=12, fontweight="bold"
)
axes[1, 1].set_xlabel("Student ID")
axes[1, 1].set_ylabel("Final Score")

plt.tight_layout()
plt.show()
```

```python

X_raw = df[["Study_Hours", "Attendance", "Assignment_Score"]].values
Y = df["Final_Score"].values

X_design = np.column_stack((np.ones(len(X_raw)), X_raw))

beta = np.linalg.inv(X_design.T @ X_design) @ X_design.T @ Y

print("Intercept (b0):", round(beta[0], 4))
print("Study Hours Coeff (b1):", round(beta[1], 4))
print("Attendance Coeff (b2):", round(beta[2], 4))
print("Assignment Score Coeff (b3):", round(beta[3], 4))
```
